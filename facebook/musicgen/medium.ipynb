# Load facebook/musicgen-medium in Google Colab
This notebook helps you easily load and run the `facebook/musicgen-medium` audio generation model in Google Colab.

```python
from transformers import pipeline
import scipy

model_id = "facebook/musicgen-medium"

synthesizer = pipeline("text-to-audio", model=model_id, device=0)

prompt = "A cheerful upbeat pop song"
music = synthesizer(prompt, forward_params={"max_new_tokens": 256})

scipy.io.wavfile.write("output.wav", rate=music["sampling_rate"], data=music["audio"])
print("Audio saved to output.wav")
```
