# Load zai-org/CogVideoX-5b in Google Colab
This notebook helps you easily load and run the `zai-org/CogVideoX-5b` video generation model in Google Colab using diffusers.

```python
import torch
from diffusers import DiffusionPipeline
from diffusers.utils import export_to_video

model_id = "zai-org/CogVideoX-5b"

pipe = DiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    variant="fp16"
)
pipe = pipe.to("cuda")
```

```python
prompt = "A dog running in the park"
video_frames = pipe(prompt, num_frames=16).frames[0]
export_to_video(video_frames, "output.mp4", fps=7)
print("Video saved to output.mp4")
```
