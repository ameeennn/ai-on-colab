# Load Kwai-Kolors/Kolors in Google Colab
This notebook helps you easily load and run the `Kwai-Kolors/Kolors` image generation model in Google Colab using diffusers.

```python
import torch
from diffusers import DiffusionPipeline

model_id = "Kwai-Kolors/Kolors"

pipe = DiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")
```

```python
prompt = "A beautiful sunset over a serene lake, digital art"
image = pipe(prompt).images[0]
image
```
